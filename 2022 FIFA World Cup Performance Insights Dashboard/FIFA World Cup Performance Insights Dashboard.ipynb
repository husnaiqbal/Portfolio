```python
from dash import Dash
from dash import dcc
from dash import html
from dash.dependencies import Input, Output
from pandas_datareader import data
import pandas as pd
import plotly.graph_objects as go
```

```python
my_dash = Dash('my dash')

df0 = pd.read_csv('/Users/husna/Downloads/BAS476/Fifa_world_cup_matches.csv')

print(df0.head())

name = 'Fifa'


team_input1 = "QATAR"
team_input2 = "ENGLAND"

style_for_label=dict(color="blue",
                     fontWeight="bold",
                     borderRadius='5px',
                     fontSize=20)
style_for_input=dict(width='200px', 
                     height='30px',
                     border='2px solid blue',
                     fontSize=20)
```

```python
# Define a plotly figure:


fig=go.Figure([go.Scatter(x= df0['team1'],
                          y= df0['number of goals team1'],
                          name = name,
                          line= dict(color="red", width=5, dash="dot")),
               #Let's put another trace on the plot
               go.Scatter(x= df0['team2'],
                          y= df0['number of goals team2'],
                          name= name,
                          line= dict(color="blue", width=5, dash="dot"))
               ])


```

```python

# Layout Components


myHeader = html.H1('Fifa Teams Comparisons') 

myLabel1  = html.Label('Team One', style=style_for_label)   
myInput1  = dcc.Input(value="QATAR",   #This value will be displayed by default    
                     id='my_input1',   #This is a variable name
                     style=style_for_input)

myLabel2  = html.Label('Team Two', style=style_for_label)   
myInput2  = dcc.Input(value = "ENGLAND",   #This value will be displayed by default    
                     id='my_input2',   #This is a variable name
                     style=style_for_input)


myGraph  = dcc.Graph(figure = fig)  # This is a static graph, it will be displayed by default  
myContainer = html.Div(id = 'my_graph', children = myGraph)


```

```python
# App layout

my_dash.layout = html.Div( children=[   myHeader,
                                       html.Div(id = 'first_team', children =[myLabel1, myInput1]),  # To make sure that one Label and one Input are displayed on one row
                                       html.Div(id = 'second_team', children =[myLabel2, myInput2]), # To make sure that one Label and one Input are displayed on one row                                       myContainer 
                                       myContainer
                                   ]
                         ) 

```
